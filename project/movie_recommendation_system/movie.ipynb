```python
## slect the columne(feature)
# genres, keywords, id, overview, title , cast, crew 

df = df[['id','title','overview','genres','keywords','cast','crew']]
```

```python
df.head()
```

```python
## find the null
df.isnull().sum()
```

```python
df.dropna(inplace=True) 
```

```python
df.isnull().sum()
```

```python
#check the duplicated
df.duplicated().sum()
```

```python
## create function for the get only list of value

def extract_values(obj):
    data = json.loads(obj)
    list = []
    for i in data:
        list.append(i.get('name'))

    return list
```

- apply the extract value functoin in the genres

```python
df.iloc[0].genres
```

```python
df['genres'] = df['genres'].apply(extract_values)
```

-  apply the extract value functoin in the keyword

```python
df['keywords'][0]
```

```python
df['keywords'] = df['keywords'].apply(extract_values)
```

```python
## extract top 3 4 dict name 
def extract_values_specific_count(obj,count=3):
    data = json.loads(obj)[0:count]
    list = []
    for i in data:
        list.append(i.get('name'))

    return list
```

-  apply the extract value functoin in the keyword

```python
df['cast'] = df['cast'].apply(extract_values_specific_count,count=3)
```

```python
df['cast'][0]
```

```python
## find direactor only
def find_direactor(obj):
    data = json.loads(obj)
    directors  = []
    for i in data:
        if i.get('job') == "Director":
            directors .append(i.get('name'))
            break
    return directors
```

```python
df['crew'] = df['crew'].apply(find_direactor)
```

```python
df['crew'][0]
```

```python
df['overview'] = df['overview'].apply(lambda x : x.split() )
```

```python
df.head()
```

```python
## remove space for word's
def removeSpace(obj):
    return [i.replace(' ','') for i in obj]
```

```python
df['genres'] = df['genres'].apply(removeSpace)
df['cast'] = df['cast'].apply(removeSpace)
df['keywords'] = df['keywords'].apply(removeSpace)
df['crew'] = df['crew'].apply(removeSpace)
```

```python
df.head()
```

```python
## concatination of the text
df['tags'] = df['overview'] + df['genres'] + df['keywords'] + df['cast'] + df['crew']
```

```python
df.head()
```

```python
new_df = df[['id','title','tags']]
```

```python
new_df.head()
```

```python
## join the tag and conver to the string
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
## lowe case convert all the text 
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
new_df.head()
```

```python
## apply staming
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return ' '.join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
#count the vectorized word
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
## use function and not model create just calculate and find the output
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity =  cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x : x[1])
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movie_list = sorted(list(enumerate(similarity[movie_index])),reverse=True,key=lambda x : x[1])

    ## not add it self movie after the select movie top 5
    for i in movie_list[1:6]:
      print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
import faiss
```

```python
vectors = cv.fit_transform(new_df["tags"]).toarray().astype("float32")
```

```python
vectors
```

```python
faiss.normalize_L2(vectors) 
```

```python
d = vectors.shape[1]               # number of features
index = faiss.IndexFlatIP(d)       # IP = inner product
index.add(vectors)                 # like fit() in KNN
```

```python
def recommend(movie_name, k=5):
    i = new_df[new_df["title"] == movie_name].index[0]
    scores, ids = index.search(vectors[i:i+1], k + 1)   # +1 because the movie itself comes first
    for j, s in zip(ids[0][1:], scores[0][1:]):
        print(new_df.iloc[j]["title"], round(float(s), 3))
```

```python
recommend("Avatar")
```
