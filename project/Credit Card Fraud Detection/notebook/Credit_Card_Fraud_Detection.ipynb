```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=["Class"])
y = df["Class"].astype(int)
```

```python
# First, reserve 15% for final testing.
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    stratify=y,
    random_state=42
)
```

```python
# Reserve approximately 15% of the complete dataset for validation.
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.15 / 0.85,
    stratify=y_train_val,
    random_state=42
)
```

```python
print("Train:", X_train.shape, y_train.sum())
print("Validation:", X_val.shape, y_val.sum())
print("Test:", X_test.shape, y_test.sum())
```

```python
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Negative:", negative_count)
print("Positive:", positive_count)
print("Class weight ratio:", scale_pos_weight)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    precision_recall_curve,
    roc_auc_score
)
import joblib
import json
```

```python
model = XGBClassifier(
    objective="binary:logistic",
    eval_metric="aucpr",
    n_estimators=600,
    learning_rate=0.05,
    max_depth=5,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=2,
    scale_pos_weight=scale_pos_weight,
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)
```

```python
model.fit(X_train, y_train)
```

```python
val_prob = model.predict_proba(X_val)[:, 1]
```

```python
precision, recall, thresholds = precision_recall_curve(
    y_val, val_prob
)
```

```python
# F1 for each threshold; final precision/recall point has no threshold.
f1_values = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)
```

```python
best_idx = int(np.argmax(f1_values))
threshold = float(thresholds[best_idx])

print("\nValidation threshold:", threshold)
print("Validation precision:", precision[best_idx])
print("Validation recall:", recall[best_idx])
print("Validation F1:", f1_values[best_idx])
```

```python
# 8. Final evaluation on the untouched test set
test_prob = model.predict_proba(X_test)[:, 1]
test_pred = (test_prob >= threshold).astype(int)

print("\nTEST CLASSIFICATION REPORT")
print(classification_report(
    y_test, test_pred,
    labels=[0, 1],
    target_names=["Normal", "Fraud"],
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_test, test_pred, labels=[0, 1]))

print("Test PR-AUC / Average precision:",
      average_precision_score(y_test, test_prob))
print("Test ROC-AUC:", roc_auc_score(y_test, test_prob))
print("Test precision:", precision_score(y_test, test_pred, zero_division=0))
print("Test recall:", recall_score(y_test, test_pred, zero_division=0))
print("Test F1:", f1_score(y_test, test_pred, zero_division=0))
```

```python
joblib.dump(model,"../model/fraud_model.joblib")
```

```python
with open("../model/config.json", "w") as f:
    json.dump({
        "threshold": threshold,
        "features": list(X.columns),
        "scale_pos_weight": scale_pos_weight
    }, f, indent=2)

print("\nModel and configuration saved.")
```
