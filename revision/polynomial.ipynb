```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26])
```

```python
plt.scatter(X,y)
plt.plot(X,y,'r')
```

```python
ploy = PolynomialFeatures(degree=2)
```

```python
X_poly = ploy.fit_transform(X)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_poly,y)
```

```python
X_range = np.linspace(X.min(), X.max(), 20).reshape(-1, 1)
```

```python
X_range_poly = ploy.transform(X_range)
```

```python
y_pred = lr.predict(X_range_poly)
```

```python
print("Intercept:", lr.intercept_)
print("Coefficients:", lr.coef_)
print("Predictions:", y_pred)
```

```python
plt.scatter(X, y, color="red", label="Actual")
plt.plot(X_range, y_pred, color="blue", label="Polynomial Regression")

plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 5, 10, 17, 26])
```

```python
degree = 2
```

```python
X_poly = np.column_stack([X**i for i in range(degree + 1)])
```

```python
X_poly
```

```python
theta = np.linalg.inv(X_poly.T @ X_poly) @ X_poly.T @ y
```

```python
print("Coefficients:", theta)
```

```python
y_pred = X_poly @ theta
```

```python
print("Predictions:", y_pred)
```

```python
plt.scatter(X, y, color="red", label="Actual")
plt.plot(X, y_pred, color="blue", label="Polynomial Regression")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```
