```python
import pandas as pd
import seaborn as sns
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score 
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

```python
## apply feature construction
```

```python
X['Family_count'] = X['SibSp'] + X['Parch'] + 1
```

```python
def CheckFamilySize(num):
    if num == 1:
        return 0
    elif num > 1 and num <= 4:
        return 1
    else:
        return 2
```

```python
X['family_size'] = X['Family_count'].apply(CheckFamilySize)
```

```python
X.head()
```

```python
new_X = X.drop(columns=['SibSp','Parch','Family_count'])
```

```python
new_X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),new_X,y,scoring='accuracy',cv=20))
```
